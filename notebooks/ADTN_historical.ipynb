# ADTN Historical Data Analysis

This notebook demonstrates fetching and analyzing ADTN (Adtran Holdings) historical data using yfinance.

```python
import sys
sys.path.insert(0, '../src')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from data.yfinance_downloader import YFinanceDownloader

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
```

```python
downloader = YFinanceDownloader(cache_dir="../data")

# Download full daily history
daily_df = downloader.download_daily_full("ADTN")
print(f"Daily data: {len(daily_df)} rows")
print(f"Date range: {daily_df['timestamp'].min()} to {daily_df['timestamp'].max()}")
daily_df.head()
```

```python
daily_df.tail()
```

```python
# Plot price history
fig, axes = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

axes[0].plot(daily_df["timestamp"], daily_df["close"], label="Close", color="blue")
axes[0].set_ylabel("Price ($)")
axes[0].set_title("ADTN Daily Close Price")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].bar(daily_df["timestamp"], daily_df["volume"], label="Volume", color="gray", alpha=0.6, width=1)
axes[1].set_ylabel("Volume")
axes[1].set_xlabel("Date")
axes[1].set_title("ADTN Daily Volume")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Basic statistics
daily_df["returns"] = daily_df["close"].pct_change()

print("=== Daily Return Statistics ===")
print(daily_df["returns"].describe())

print(f"\nAnnualized Volatility: {daily_df['returns'].std() * (252**0.5):.2%}")
print(f"Annualized Return (CAGR): {((daily_df['close'].iloc[-1] / daily_df['close'].iloc[0]) ** (252/len(daily_df)) - 1):.2%}")
```

```python
# Download recent intraday data
intraday_df = downloader.download_intraday_recent("ADTN", days=30)
print(f"Intraday data: {len(intraday_df)} rows")
print(f"Date range: {intraday_df['timestamp'].min()} to {intraday_df['timestamp'].max()}")
intraday_df.head()
```

```python
# Plot intraday
fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

axes[0].plot(intraday_df["timestamp"], intraday_df["close"], label="Close", color="blue")
axes[0].set_ylabel("Price ($)")
axes[0].set_title("ADTN Intraday (Hourly) - Last 30 Days")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].bar(intraday_df["timestamp"], intraday_df["volume"], label="Volume", color="gray", alpha=0.6, width=0.02)
axes[1].set_ylabel("Volume")
axes[1].set_xlabel("Date")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
